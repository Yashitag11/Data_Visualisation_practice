```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt
import numpy as np

# scatter plot
mark1 = [96,97,98,99,89,90,91]
mark2 = [90,89,79,98,96,95,94]
# correlation - relationship between 2 variables
corr = np.corrcoef(mark1,mark2)[0][1] #corr of x with y
print(f"correlation: {corr:.2f}")
plt.scatter(mark1, mark2 ,s=100,color='pink', marker='o')
# trend line
z = np.polyfit(mark1,mark2,1)
p = np.poly1d(z)
plt.plot(mark1,p(mark1), color="red",linestyle='--', label='trend line')
plt.title('Scatter plot')
plt.xlabel('marks of student 1')
plt.ylabel('marks of student2')
plt.show()
```

```python
# heatmap
# basic
data = np.array([[1.0, 0.8, 0.3],
                 [0.8, 1.0, 0.5],
                 [0.3, 0.5, 1.0]])

plt.imshow(data, cmap='coolwarm')
plt.colorbar()
plt.title('Basic Heatmap')
plt.show()
# the scale is given beside you just have interpret that using logic and correlation concepts
# like here dark red - 1 -> prefect light red - 0.8 -> strong and dark blue - 0.3 -> weak
```

```python
# styled
data = np.array([[1.0, 0.8, 0.3],
                 [0.8, 1.0, 0.5],
                 [0.3, 0.5, 1.0]])
# this matrix shows the correlation value already
subjects = ['Math', 'Science', 'English']

plt.imshow(data, cmap='coolwarm')
plt.colorbar(label='Correlation Value')

plt.xticks([0, 1, 2], subjects) #name of subjects beside them
plt.yticks([0, 1, 2], subjects)

plt.title('Subject Correlation Heatmap')
plt.show()
```

```python
# with numbers inside cell
data = np.array([[1.0, 0.8, 0.3],
                 [0.8, 1.0, 0.5],
                 [0.3, 0.5, 1.0]])

subjects = ['Math', 'Science', 'English']

plt.imshow(data, cmap='coolwarm')
plt.colorbar(label='Correlation')

# Add numbers inside each cell
for i in __builtins__.range(3):
    for j in __builtins__.range(3):
        plt.text(j, i, f'{data[i,j]:.2f}',
                 ha='center', va='center',
                 fontsize=12, fontweight='bold',
                 color='black')

plt.xticks([0, 1, 2], subjects)
plt.yticks([0, 1, 2], subjects)
plt.title('Heatmap with Values')
plt.show()
```

```python
# heatmap correlation
math    = [80, 85, 90, 78, 92, 88, 76]
science = [78, 82, 88, 75, 90, 85, 74]
english = [60, 55, 65, 70, 58, 62, 68]
hindi   = [70, 72, 75, 68, 78, 74, 66]

# Stack data and find correlation
data    = np.array([math, science, english, hindi])
corr    = np.corrcoef(data)

subjects = ['Math', 'Science', 'English', 'Hindi']

plt.imshow(corr, cmap='coolwarm')
plt.colorbar(label='Correlation')

# Add values inside cells
for i in __builtins__.range(4):
    for j in __builtins__.range(4):
        plt.text(j, i, f'{corr[i,j]:.2f}',
                 ha='center', va='center',
                 fontsize=11, fontweight='bold')

plt.xticks([0, 1, 2, 3], subjects)
plt.yticks([0, 1, 2, 3], subjects)
plt.title('Subject Correlation Heatmap')
plt.show()
```

```python
# basic density
from scipy.stats import gaussian_kde

marks = [45, 55, 60, 70, 72, 75, 78,
         80, 82, 85, 88, 90, 92, 95, 98]

# Step 1 - Create KDE
kde = gaussian_kde(marks)

# Step 2 - Create smooth x values
x = np.linspace(min(marks), max(marks), 200)

# Step 3 - Plot
plt.plot(x, kde(x), color='blue', linewidth=2)
plt.title('Density Plot of Marks')
plt.xlabel('Marks')
plt.ylabel('Density')
plt.grid(True)
plt.show()
```

```python
# filling area in between
marks = [45, 55, 60, 70, 72, 75, 78,
         80, 82, 85, 88, 90, 92, 95, 98]

kde = gaussian_kde(marks)
x   = np.linspace(min(marks), max(marks), 200)

# Plot line
plt.plot(x, kde(x), color='blue', linewidth=2)

# Fill area under curve
plt.fill_between(x, kde(x), alpha=0.3, color='blue')

plt.title('Density Plot with Fill')
plt.xlabel('Marks')
plt.ylabel('Density')
plt.grid(True)
plt.show()
```

```python
# comparing 2 groups
class_a = [55, 60, 65, 70, 75, 80, 85, 90]
class_b = [70, 72, 75, 78, 82, 85, 88, 95]

# Class A kde
kde_a = gaussian_kde(class_a)
x_a   = np.linspace(min(class_a), max(class_a), 200)

# Class B kde
kde_b = gaussian_kde(class_b)
x_b   = np.linspace(min(class_b), max(class_b), 200)

# Plot both
plt.plot(x_a, kde_a(x_a), color='blue', linewidth=2, label='Class A')
plt.fill_between(x_a, kde_a(x_a), alpha=0.3, color='blue')

plt.plot(x_b, kde_b(x_b), color='red', linewidth=2, label='Class B')
plt.fill_between(x_b, kde_b(x_b), alpha=0.3, color='red')

plt.title('Class A vs Class B')
plt.xlabel('Marks')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.show()

# conclusion
# Looking at this density plot Class B has performed better than Class A overall. Class B peak is around 78 marks while Class A peak is around 70 marks showing Class B students scored higher on average.
# Class A has a wider and flatter curve starting from 55 marks showing more variation and weaker students. Class B has a taller and narrower curve starting from 70 marks showing students are more consistent.
# The purple overlapping region between 70 and 90 shows both classes have students in this range. Both distributions follow a normal bell curve shape with no skewness.
# Overall Class B is the stronger and more consistent class compared to Class A
```

```python
# area chart
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales  = [200, 250, 230, 300, 280, 350]

plt.fill_between(months, sales,
                 alpha=0.4, color='blue')
plt.plot(months, sales,
         color='blue', linewidth=2)

plt.title('Monthly Sales Area Chart')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
# comparing 2 areas
months    = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
product_a = [200, 250, 230, 300, 280, 350]
product_b = [150, 180, 200, 220, 260, 300]

# Product A
plt.fill_between(months, product_a,
                 alpha=0.4, color='blue')
plt.plot(months, product_a,
         color='blue', linewidth=2,
         marker='o', label='Product A')

# Product B
plt.fill_between(months, product_b,
                 alpha=0.4, color='red')
plt.plot(months, product_b,
         color='red', linewidth=2,
         marker='s', label='Product B')

plt.title('Product A vs Product B Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.show()
```
