```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
# hue is a color coding - produce element swith different colors, eg in our it is divided wrt time
# time coding variable 
# style = will show different variables x,y using a symbol
# size = is used to differentiate what size of dot is -> which value is present where
# sizes = object that determines what size is to be chosen (min,max) tuple (predefining the size od the circles)
# col = it shows 2 different graph for each value in mentioned column
```

```python
# scatter plot
sns.scatterplot(x='total_bill', y='tip',data = tips)
```

```python
# relationship plot
sns.relplot(x='total_bill', y='tip',hue='time',data = tips)
```

```python
sns.relplot(x='total_bill', y='tip',data = tips, style='smoker')
```

```python
sns.relplot(x='total_bill', y='tip',data = tips,hue='time',style='smoker')
# this graph shows the time for lunch and dinners through hue and color coding while smoker values yes and no through style
```

```python
sns.relplot(x='total_bill', y='tip',data = tips, size='size')
```

```python
sns.relplot(x='total_bill', y='tip',data = tips,size='size',sizes=(15,500))
```

```python
sns.relplot(x='total_bill', y='tip',data = tips,kind='line')
# kind = helps in converting to line plot
```

```python
sns.relplot(x='size', y ='tip',data=tips, hue='time',col='smoker',kind='line')
```

```python
# multiple plot 
# Seaborn to create a multi-panel line chart showing how tips change based on the total bill for male customers only, 
# split into separate subplots by the day of the week and colored by the time of day (Lunch/Dinner).
sns.relplot(x='total_bill',y='tip',hue='time',col='day',col_wrap=5,height=3,aspect=.75,linewidth=2.5,kind='line',
                                                  data=tips.query('sex=="Male"'))
# height and aspect tells height and width set for each subplots
# col_wrap=5 ensures the subplots wrap to a new line after reaching 5 columns.
# hue='time'Adds a third dimension by coloring the lines and legend items differently depending on the time of day
```

```python
g = sns.relplot(x='total_bill',y='tip',hue='time',col='day',col_wrap=5,height=3,aspect=.75,linewidth=2.5,kind='line',
                                                  data=tips.query('sex=="Male"'))
g.fig.subplots_adjust(wspace=.5,hspace=.5) #adding space between subplotted graph
```

```python
sns.histplot(x='total_bill', y='tip',data = tips,hue='time',bins=5,legend=True)
# ax.legend('Tips V/S Total_bill')
```

```python
sns.catplot(data=tips,x='sex',y='day')
```

<b>representing date axis <b>

```python
df = pd.DataFrame(dict(time=pd.date_range('2017-1-1',periods=600),
                      value=np.random.randn(600).cumsum()))
g = sns.relplot(x='time', y='value',kind='line',data=df)
g.fig.autofmt_xdate() #it rotates data labels so that tehy don't overlap
# periods = total no. of date needed
```

<b> CATEGORICAL DATA VISUALISATION <b>

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
# sns.catplot(x='sex',y='tip',data=tips)
# sns.catplot(x='sex',y='tip',data=tips, jitter=True) outliers are esily identifiable
sns.catplot(x='sex',y='tip',data=tips, jitter=False) #outliers are removed
```

```python
# Categorical scatterplots:

# - :func:`stripplot` (with `kind="strip"`; the default)
# - :func:`swarmplot` (with `kind="swarm"`)

# Categorical distribution plots:

# - :func:`boxplot` (with `kind="box"`)
# - :func:`violinplot` (with `kind="violin"`)
# - :func:`boxenplot` (with `kind="boxen"`)

# Categorical estimate plots:

# - :func:`pointplot` (with `kind="point"`)
# - :func:`barplot` (with `kind="bar"`)
# - :func:`countplot` (with `kind="count"`)
```

```python
tips.describe()
```

```python
sns.catplot(x='sex',y='tip',data=tips, kind='swarm',size=4,hue='sex')
# color coding the sex
# size=4 is to remove warning
```

```python
sns.catplot(x='sex',y='tip',data=tips, kind='box',hue='sex')
# for horizontal just change the axis values
```

```python
sns.catplot(x='sex',y='tip',data=tips, kind='violin',hue='smoker')
```

```python
sns.catplot(x='sex',y='tip',data=tips, kind='boxen',hue='day')
```

```python
# sns.catplot(x='sex',data=tips, kind='count',hue='sex')
sns.catplot(x='tip',data=tips, kind='count',hue='sex')
```

```python
sns.catplot(x='sex',y='tip',data=tips, kind='point',hue='smoker')
```
